# Data Cleaning Pipeline

This notebook covers the essential steps for cleaning and preparing data for analysis.

1) Import Libraries

```python
import pandas as pd
```

2) Load Dataset

```python
df = pd.read_csv("data/clean_data.csv")  
df.head()
```

3) Dataset Overview

```python
print("Shape:", df.shape)
df.info()
```

```python
df.describe()
```

4) Column Names

```python
df.columns
```

5) Check Missing Values

```python
missing_values = df.isnull().sum()
print(missing_values)
```

6) Handle Missing Values

```python
df = df.dropna()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

7) Check for Duplicates

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

8) Remove Duplicates

```python
df = df.drop_duplicates()
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```
